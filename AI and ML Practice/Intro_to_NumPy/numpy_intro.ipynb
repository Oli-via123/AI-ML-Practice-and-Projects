Introduction to NumPy
NumPy [Numerical Python] is a powerful library for numerical and scientific computing in python .
 NumPy Arrays also known as "ndarray": is the array object in NumPy and can be multi-dimensional and also fixed type
"Arrays" : are like a list but faster and supports mathematical operations directly.

Reshaping Arrays
Defining an Array and Reshaping the arrays

```python
arr_def = np.array([1,2,3,4,5,6])
reshaped_arr =arr_def.reshape(2,3)
print (reshaped_arr)
```

Flatten Multi-dimensional arrays to 1 dimension

```python
flattened_arr = reshaped_arr.flatten()
print(flattened_arr)
print(flattened_arr.shape)
```
