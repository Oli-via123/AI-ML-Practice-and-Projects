AI Grade Automation And Feedback Generation

```python
# Define a function that assign grade based on score, attendance, submission.
def grade_assign (score, attendance, submitted) :
        if score >= 80 :
            grade = "A"
        elif score >= 60 :
            grade = "B"
        elif score >= 50 :
            grade = "C"
        elif score >= 45 :
            grade = "D"
        elif score >= 40 :
            grade = "E"
        else :
            grade = "F"
            # Attendance rule 
        if attendance < 70 and grade == "A" :
                grade = "B"
                # Apply submission rule :
        if submitted == "no" and grade in ["A", "B"] :
                    grade = "C" 
        return grade            
    
```

```python
grade_assign(90, 70, "yes")
```

```python
print(grade_assign (90, 70, "yes"))
```

```python
# Create a function that generate feedback
def feedback_gen(grade) :
     feedback = { 
                 "A":"Congratulatins! This is an excellent performance.",
"B" :"Very Good! Asin higher.", 
"C" :"Good work! You can do better.",
"D" :"You passed but improvement is needed.",
"E" :"Fair result You  need to work harder for a better result.",
"F" :"Poor result so see me privately as a matter of urgency "}
     return feedback[grade]
```

```python
feedback_gen ("A")
```

```python
# User Input
name = input ("student name:")
score = int (input("score(0-100):"))
attendance = int(input("Attendance percentage:"))
submitted = input("Assignment submitted(yes/no):")
grade = grade_assign(score, attendance, submitted)
feedback = feedback_gen(grade)
status = "PASS" if grade != "F" else "FAIL"
print("~"*33)
print(f"student:{name}")
print(f"grade:{grade}")
print(f"status:{status}")
print(f"feedback:{feedback}")
```
